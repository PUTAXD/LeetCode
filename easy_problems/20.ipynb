```python
from typing import List
s = "()[]{}"
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        pair = {')' : '(','}' : '{', ']' : '['}

        if (len(s)%2) != 0: 
            return False
        
        for c in s : 
            if c in pair : 
                if pair[c] == stack[-1] :
                    stack.pop()
                else:
                    return False
            else:
                stack.append(c)
        
        return True
```

```python
Solution.isValid(Solution(), s)
```
