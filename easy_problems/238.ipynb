```python
from typing import List
nums = [1,2,3,4]
```

```python
class Solution:
    def productExceptSelfBruteForce(self, nums: List[int]) -> List[int]:
        res = []
        for i in range(len(nums)) : 
            sum = 1
            for j in range(len(nums)):
                if j == i :
                    continue
                sum = (nums[j]*sum)
            res.append(sum)
        return res
    
    def another(self, nums: List[int]) -> List[int]:
        res = []
        prod = 1
        for num in nums:
            prod *= num
        for num in nums:
            if num == 0 :
                continue
            res.append(int(prod/num))
        
        return res 
```

```python
Solution.productExceptSelfBruteForce(Solution(), nums)
```

```python
Solution.another(Solution(), nums)
```
