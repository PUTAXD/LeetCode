```python
from typing import List
nums = [1,2,3,4]
```

```python
class Solution:
    def productExceptSelfBruteForce(self, nums: List[int]) -> List[int]:
        res = []
        for i in range(len(nums)) : 
            sum = 1
            for j in range(len(nums)):
                if j == i :
                    continue
                sum = (nums[j]*sum)
            res.append(sum)
        return res
    
    def another(self, nums: List[int]) -> List[int]:
        res = []
        prod = 1
        for num in nums:
            prod *= num
        for num in nums:
            if num == 0 :
                res.append
            res.append(int(prod/num))
        
        return res 
    
    def productExceptSelfPrefixSufix(self, nums: List[int]) -> List[int]:
        n = len(nums)
        res = []
        prefix = [0] * n 
        suffix = [0] * n 
        
        prefix[0] = suffix[n-1] = 1
        for i in range(1, n) : 
            prefix[i] = nums[i-1] * prefix[i-1]
        for i in range(n-2, -1,-1):
            suffix[i] = nums[i+1] * suffix[i+1]

        for i in range(n) : 
            res.append(suffix[i] * prefix[i])
        
        return res
```
