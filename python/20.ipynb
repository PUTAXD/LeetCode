```python
from typing import List
s = "()[]{}]"
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        pair = {')' : '(','}' : '{', ']' : '['}

        for c in s : 
            if c in pair : 
                if stack and pair[c] == stack[-1] :
                    stack.pop()
                else:
                    return False
            else:
                stack.append(c)

        if stack: return False
        else : return True
```

```python
Solution.isValid(Solution(), s="){")
```
